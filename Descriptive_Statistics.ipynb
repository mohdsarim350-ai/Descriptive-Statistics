#Descriptive Statistics

1. Understanding Central Tendency.

   - Data: [10, 12, 11, 15, 14, 13, 12]

   - The most representative value of the weekly sales is 12.

   - This is because 12 occurs more frequently than other values and lies near the center of the dataset, making it a good representation of typical weekly sales.

2. Mean in Real Life.

    - [12, 15, 14, 16, 18, 20, 19]

    - The mean score is 16.29.

    - This indicates that the overall performance of the class is fairly good, with most students scoring around this average value.

3. Mode in Real Life.

   - Shoe sizes: [7, 8, 9, 8, 8, 10, 7, 9]

   - the mode is 8.

   - This information is useful for the store manager because it shows that size 8 is the most frequently sold shoe size and should be stocked in higher quantity.

4. Median in Real Life.

    - Car Price: [8000, 9500, 10200, 11000, 50000]

    - the median price is 10,200.

    - The median is a better measure than the mean is this case because the very high price of 50,000 is an outlier and would distort the average.

5. Dispersion Introduction.

   - Time taken (minutes): [25, 30, 27, 35, 40]

   - The range is 15 minutes.

   - This shows that the student's puzzle-solving time varies by 15 minutes, indicating noticeabel variation in performance.

6. Range in Action.

   - Apple weights (Kg): [100, 105, 98, 110, 120]

   - The range is 22 Kg.

   - This hepls the farmer understand the variation in apple weights and plan packaging accordingly.

7. Variance for Decision-Making.

   - Company A variance = 6

   - Company B variance = 15

   - Company A is more consistent.

   - Lower variance indicates less variation in delivery delays, which means more reliable service.

8. Standard Deviation in Context.

   - Coin A standard deviation = $30

   - Coin B standard deviation = $120

   - Coin B is riskier to invest in.

   - A higher standard deviation indicates greater price fluctuations and higher investment risk.

9. Combining Measures.

   - Electricity usage (kWh): [400, 420, 390, 450, 410]

   - the mean electricity usage is 414 kWh, and the standard deviation is approximately 21.6.

   - together, these values indicate that the family's energy usage is relatively stabel with moderate variation.
